```python
class ProZero(nn.Module):
    def __init__(self):
        super(ProZero, self).__init__()
        self.conv1 = nn.Conv2d(1, 8, kernel_size=3, padding=1)

        self.p_conv = nn.Conv2d(8, 2, kernel_size=1)
        self.p_fc1 = nn.Linear(84,4*42)
        self.p_fc2 = nn.Linear(4*42, 7) # Output: 9 probabilities

        self.v_conv = nn.Conv2d(8, 1, kernel_size=1)
        self.v_fc1 = nn.Linear(42, 2*42) # Small dense layer
        self.v_fc2 = nn.Linear(2*42, 3) # Output: Scalar [-1, 1]

        self.r_conv = nn.Conv2d(8, 1, kernel_size=1)
        self.r_fc1 = nn.Linear(42, 2*42) # Small dense layer
        self.r_fc2 = nn.Linear(2*42, 3) # Output: Scalar [-1, 1]

    def forward(self, x):
        x = x.view(-1, 1, 6, 7)
        x = F.relu(self.conv1(x))

        p = self.p_conv(x)
        p = p.view(-1, 2 * 6 * 7)
        p = F.relu(self.p_fc1(p))
        p = self.p_fc2(p)

        v = self.v_conv(x)
        v = v.view(-1, 1 * 6 * 7)
        v = F.relu(self.v_fc1(v))
        v = self.v_fc2(v)

        r = self.r_conv(x)
        r = r.view(-1, 1 * 6 * 7)
        r = F.relu(self.r_fc1(r))
        r = self.r_fc2(r)

        return p, v ,r
```

```python
def winner_prob(children):
    P = np.array([child.search_q for child in children.values()])
    # invert wdl
    p0 = P[:, 2]
    p1 = P[:, 1]
    p2 = P[:, 0]
    lt1 = p0 + 1e-12
    lt2 = p0 + p1 + 1e-12
    prod_lt1 = np.prod(lt1)
    prod_lt2 = np.prod(lt2)
    win = p1 * (prod_lt1 / lt1) + p2 * (prod_lt2 / lt2)
    if win.sum() == 0:
      win =  np.ones_like(win)
    win /= win.sum()
    return win
```

```python
def winner_prob_upper(children):
    P = np.array([child.search_q for child in children.values()])
    # invert wdl
    p0 = P[:, 2]
    p1 = P[:, 1]
    p2 = P[:, 0]
    lt1 = p0 + 1e-12
    lt2 = p0 + p1 + 1e-12
    lt3 = p0 + p1 + p2
    prod_lt1 = np.prod(lt1)
    prod_lt2 = np.prod(lt2)
    prod_lt3 = np.prod(lt3)
    win = p0 * (prod_lt1 / lt1) + p1 * (prod_lt2 / lt2) + p2 * (prod_lt3 / lt3)
    if win.sum() == 0:
      win =  np.ones_like(win)
    win /= win.sum()
    return win
```

```python
class Node():
  def __init__(self):
    self.ldw = []
    self.search_q = []
    self.r = []
    self.target_q = []
    self.target_r = None
    self.legal = []
    self.p = {}
    self.n = 0
    self.children = {}
    self.parents = []
    self.explored = False
    self.obs = None
```

```python
def get_obs(state):
  # 1. Get the observation, assuming it is a native Python list
  obs_list = state.observation_tensor()

  # 2. Extract the segments as Python lists
  p0_plane = obs_list[42:84]
  p1_plane = obs_list[84:126]

  # 3. Convert the segments to PyTorch tensors for vectorized subtraction
  p0_tensor = torch.tensor(p0_plane, dtype=torch.float32)
  p1_tensor = torch.tensor(p1_plane, dtype=torch.float32)

  player = state.current_player()

  if player == 0:
    # Player 0's perspective: P0 is 1, P1 is -1
    # Result: 9-element tensor
    return p0_tensor - p1_tensor
  else:
    # Player 1's perspective: P1 is 1, P0 is -1
    # Result: 9-element tensor
    return p1_tensor - p0_tensor
```

```python
class MCTS():
  def __init__(self,model):
    self.game = pyspiel.load_game("connect_four")
    self.state = self.game.new_initial_state()
    self.model = model.to(device)
    self.model.eval()
    self.node_dict = {}
    self.terminal_set = set()

  def run_game(self,playouts=100):

    self.state = self.game.new_initial_state()
    self.node_dict = {}
    self.terminal_set = set()
    observation = []
    policy = []
    value = []
    reward_with_weight = []
    policy_value_weight = []
    legal = []

    def forward(node):
      node.n += 1
      if self.state.is_terminal() :
        if node.explored == True:
          return None , None
        node.explored = True
        rewards = self.state.returns()
        if rewards[0] == 0.0 and rewards[1] == 0.0:
              node.ldw = (0.0 , 1.0 ,0.0)
              node.search_q = (0.0 , 1.0 ,0.0)
              node.r = (0.0 , 1.0 ,0.0)
              node.target_q = (0.0 , 1.0 ,0.0)
              node.target_r = (0.0 , 1.0 ,0.0)
              return node, 'draw'
        else:
              node.ldw = (1.0 , 0.0 ,0.0)
              node.search_q = (1.0 , 0.0 ,0.0)
              node.r = (1.0 , 0.0 ,0.0)
              node.target_q = (1.0 , 0.0 ,0.0)
              node.target_r = (1.0 , 0.0 ,0.0)
              return node, 'x' if self.state.current_player()==1 else 'o'

      elif node.explored == False :
        node.explored = True
        node.legal = self.state.legal_actions()
        x = torch.tensor(get_obs(self.state), dtype=torch.float32, device=device)
        x = x.unsqueeze(0)
        policy , value ,reward = self.model(x)

        node.ldw = F.softmax(value, dim=-1)[0].tolist()
        node.search_q = node.ldw.copy()
        node.r = F.softmax(reward, dim=-1)[0].tolist()
        node.target_q = node.r.copy()
        self.node_dict[str(self.state)] = node
        child_probs = []

        for idx in node.legal:
          child_probs.append(policy[0][idx].item())
        child_probs = torch.tensor(child_probs, device=device)
        child_probs = torch.softmax(child_probs, dim=0).cpu().numpy()
        node.p = {node.legal[idx] : child_probs[idx] for idx in range(len(node.legal))}

        return node , None

      else:
        # traverse
        move = random.choices(list(node.p.keys()), weights=list(node.p.values()), k=1)[0]

        player = self.state.current_player()
        curr_state = self.state.clone()
        self.state.apply_action(move)
        if str(self.state) not in self.node_dict :
          node.children[move] = Node()
          node.children[move].parents.append((node,move))
          if self.state.current_player() == 0 or self.state.current_player() == 1:
            node.children[move].obs = get_obs(self.state)
        else:
          node.children[move] = self.node_dict[str(self.state)]

        leaf , terminal = forward(node.children[move])

        # self.state.undo_action(player,move)
        self.state = curr_state

        return leaf , terminal

    def add_to_set(node,traversal_set):
      if node in traversal_set:
        return
      traversal_set.add(node)
      for parent,move in node.parents:
        add_to_set(parent,traversal_set)
      return

    def forward_update(node,traversal_set):

      traversal_set.remove(node)
      for idx , child in node.children.items() :
        if child in traversal_set:
          forward_update(child,traversal_set)

      if not node.children:
        return
      D = np.array([explored.search_q for explored in node.children.values()]+[[node.ldw[2],node.ldw[1],node.ldw[0]]])
      R = np.array([explored.target_q for explored in node.children.values()]+[[node.r[2],node.r[1],node.r[0]]])
      explored_p = sum([node.p[explored] for explored in node.children.keys()])
      scaled_probs = winner_prob(node.children)*explored_p
      for i,idx in enumerate(node.children.keys()):
        node.p[idx] = scaled_probs[i]
      P = np.concatenate((scaled_probs, [max(0,1-explored_p)]))
      node.search_q = np.sum(P[:, None] * D, axis=0)
      node.target_q = np.sum(P[:, None] * R, axis=0)

      node.search_q = [node.search_q[2],node.search_q[1],node.search_q[0]]
      node.target_q = [node.target_q[2],node.target_q[1],node.target_q[0]]

      return

    def final_update(node):
      self.terminal_set.remove(node)
      if node.target_r is not None:
        return

      node.target_r = [0,0,0]

      for idx , child in node.children.items() :
        if child in self.terminal_set:
          final_update(child)

      # maxp = max(node.p.values())
      upper_p = winner_prob_upper(node.children)
      i = 0
      for idx , child in node.children.items() :
        if child.target_r is None:
          continue
        # weight = node.p[idx]/maxp
        weight = upper_p[i]
        i+=1
        node.target_r[0] += weight*child.target_r[2]
        node.target_r[1] += weight*child.target_r[1]
        node.target_r[2] += weight*child.target_r[0]

      #observation
      observation.append(node.obs)
      #policy
      policy_ = np.zeros(self.state.num_distinct_actions())
      for idx in node.children.keys() :
        policy_[idx] = node.p[idx]
      policy_ /= sum(policy_)
      policy.append(policy_)
      #value
      value.append(node.target_q)
      #reward
      reward_with_weight.append(node.target_r)
      #node_count
      policy_value_weight.append(sqrt(node.n))
      # policy_value_weight.append(node.n)
      #legal
      legal.append(node.legal)

      return

    def playout(node,root):
      for i in range(playouts):
        leaf,terminal = forward(node)
        if leaf is None :
          continue
        traversal_set = set()
        add_to_set(leaf,traversal_set)
        if terminal is not None:
          self.terminal_set |= traversal_set
        forward_update(root,traversal_set)

    curr = Node()
    curr.obs = get_obs(self.state)
    root = curr
    while self.state.current_player() == 0 or self.state.current_player() == 1 :
      playout(curr,root)
      moves = list(curr.children.keys())
      moves_weight = [curr.p[move] for move in moves]
      move = random.choices(moves , weights = moves_weight)[0]
      # move = max(moves, key=lambda k: curr.p[k])
      curr = curr.children[move]
      self.state.apply_action(move)
    final_update(root)

    return {
        "observation": observation,
        "policy": policy,
        "value": value,
        "reward": reward_with_weight,
        "weight": policy_value_weight,
        "legal": legal
    }
```

```python
class ReplayBuffer(Dataset):
    def __init__(self, capacity=5000):
        self.buffer = deque(maxlen=capacity)

    def save_game(self, game_data):
        """
        Ingests the dictionary returned by MCTS.run_game()
        """
        observations = game_data["observation"]
        policies = game_data["policy"]
        values = game_data["value"]
        rewards = game_data["reward"] # These are raw sums [a,b,c]
        weights = game_data["weight"] # sqrt(n)
        legals = game_data["legal"]   # List of lists [[0,1], [4,5]...]

        for o, p, v, r, w, l in zip(observations, policies, values, rewards, weights, legals):

            # Create Binary Mask for Legal Moves
            mask = np.zeros(7, dtype=np.float32)
            if len(l) > 0:
                mask[l] = 1.0

            # Convert to lightweight numpy types for storage
            self.buffer.append((
                np.array(o, dtype=np.float32),
                np.array(p, dtype=np.float32),
                np.array(v, dtype=np.float32),
                np.array(r, dtype=np.float32),
                np.float32(w),
                mask
            ))

    def __len__(self):
        return len(self.buffer)

    def __getitem__(self, idx):
        return self.buffer[idx]

class AlphaZeroTrainer:
    def __init__(self, model, optimizer):
        self.model = model
        self.optimizer = optimizer
        self.device = device

    def train_epoch(self, dataloader):
        self.model.train()

        total_loss = 0
        total_p = 0
        total_v = 0
        total_r = 0
        batches = 0

        for batch_idx, (obs, target_p, target_v, raw_target_r, p_v_weight, legal_mask) in enumerate(dataloader):
            # print(obs, target_p, target_v, raw_target_r, p_v_weight, legal_mask)
            # Move to GPU
            obs = obs.to(self.device)
            target_p = target_p.to(self.device)
            target_v = target_v.to(self.device)
            raw_target_r = raw_target_r.to(self.device)
            p_v_weight = p_v_weight.to(self.device).unsqueeze(1)
            legal_mask = legal_mask.to(self.device)

            # --- Forward Pass ---
            pred_p, pred_v, pred_r = self.model(obs)

            # --- 1. Policy Loss (Masked & Weighted) ---
            # Mask illegal moves with -1e9 (effectively 0 probability after softmax)
            pred_p_masked = pred_p.clone()
            # print(legal_mask)
            # print(pred_p_masked)
            pred_p_masked[legal_mask == 0] = -1e9

            log_probs = F.log_softmax(pred_p_masked, dim=-1)

            # Safety: Clamp weights to prevent explosion
            # p_v_weight = torch.clamp(p_v_weight, min=1e-5, max=50.0)

            # Cross Entropy: - sum(target * log(pred))
            p_loss_raw = -(target_p * log_probs).sum(dim=-1, keepdim=True)
            p_loss = (p_loss_raw * p_v_weight).mean()

            # --- 2. Value Loss (Weighted) ---
            # MSE on Softmaxed values
            v_probs = F.softmax(pred_v, dim=-1)
            v_loss_raw = F.mse_loss(v_probs, target_v, reduction='none').sum(dim=-1, keepdim=True)
            v_loss = (v_loss_raw * p_v_weight).mean()

            # --- 3. Reward Loss (Dynamic Normalization) ---
            # Normalize targets: [10, 5, 0] -> [0.66, 0.33, 0]
            r_sum = raw_target_r.sum(dim=1, keepdim=True) + 1e-6
            target_r_probs = raw_target_r / r_sum

            # Dynamic weight: sqrt(sum)
            r_weight = torch.sqrt(r_sum)
            # r_weight = r_sum
            # r_weight = torch.clamp(r_weight, min=1e-5, max=50.0)

            # MSE on Softmaxed rewards
            r_probs = F.softmax(pred_r, dim=-1)
            r_loss_raw = F.mse_loss(r_probs, target_r_probs, reduction='none').sum(dim=-1, keepdim=True)
            r_loss = (r_loss_raw * r_weight).mean()

            # --- Optimization ---
            loss = p_loss + v_loss + r_loss

            # Check for NaNs before backward to avoid crashing the run
            if torch.isnan(loss):
                print(f"Skipping Batch {batch_idx}: Loss is NaN")
                continue

            self.optimizer.zero_grad()
            loss.backward()

            # *** CRITICAL: Gradient Clipping ***
            # This prevents the 'ValueError: finite' and NaNs you saw earlier
            # torch.nn.utils.clip_grad_norm_(self.model.parameters(), max_norm=1.0)

            self.optimizer.step()

            total_loss += loss.item()
            total_p += p_loss.item()
            total_v += v_loss.item()
            total_r += r_loss.item()
            batches += 1

        if batches == 0: return {}
        return {
            "loss": total_loss / batches,
            "p_loss": total_p / batches,
            "v_loss": total_v / batches,
            "r_loss": total_r / batches
        }
```

```python

# --- Configuration ---
CHECKPOINT_PATH = "/content/model_iter_00.pth" # <--- The file you want to resume from

# Training Hyperparameters
# input_dim = 27
NUM_ITERATIONS = 100     # How many MORE iterations to run
GAMES_PER_ITER = 25      # Increased slightly to get more variety
PLAYOUTS = 200
BATCH_SIZE = 64
EPOCHS = 1

# --- Init System ---
print(f"Initializing on {device}...")
model = ProZero().to(device)

optimizer = optim.Adam(model.parameters(), lr=1e-4)

replay_buffer = ReplayBuffer(capacity=10000)

trainer = AlphaZeroTrainer(model, optimizer)
mcts = MCTS(model)


# --- Load Checkpoint Logic ---
start_iter = 0
if os.path.exists(CHECKPOINT_PATH):
    print(f">> Loading checkpoint: {CHECKPOINT_PATH}")
    try:
        model.load_state_dict(torch.load(CHECKPOINT_PATH, map_location=device))
        print(">> Model weights loaded successfully.")

        # Try to guess the iteration number from the filename
        match = re.search(r"iter_(\d+)", CHECKPOINT_PATH)
        if match:
            start_iter = int(match.group(1))
            print(f">> Resuming logs from Iteration {start_iter + 1}")
    except Exception as e:
        print(f"!! Failed to load checkpoint: {e}")
        print("!! Starting from scratch.")
else:
    print(f">> No checkpoint found at {CHECKPOINT_PATH}. Starting from scratch.")

# --- Training Loop ---
for iteration in range(start_iter, start_iter + NUM_ITERATIONS):
    print(f"\n--- Iteration {iteration+1} ---")

    # A. Self-Play
    model.eval()
    print(f">> Self-playing ({GAMES_PER_ITER} games)...")
    for i in range(GAMES_PER_ITER):
        game_data = mcts.run_game(playouts=PLAYOUTS)
        replay_buffer.save_game(game_data)

    print(f"Buffer size: {len(replay_buffer)}")

    # B. Training
    if len(replay_buffer) > BATCH_SIZE:
        print(">> Training...")
        dataloader = DataLoader(replay_buffer, batch_size=BATCH_SIZE, shuffle=True)

        for epoch in range(EPOCHS):
            metrics = trainer.train_epoch(dataloader)
            if metrics:
                print(f"   Ep {epoch}: Loss={metrics['loss']:.4f} (P={metrics['p_loss']:.3f}, V={metrics['v_loss']:.3f}, R={metrics['r_loss']:.3f})")
            else:
                print(f"   Ep {epoch}: Failed (NaNs)")

    if (iteration + 1) % 10 == 0:
        save_name = f"model_iter_{iteration+1}.pth"
        torch.save(model.state_dict(), save_name)
        print(f">> Saved checkpoint: {save_name}")
```

```python
NUM_ROWS = 6
NUM_COLS = 7
NUM_ACTIONS =
BOARD_SIZE = NUM_ROWS * NUM_COLS
# ------------------------------

# ==========================================
# 1. VISUALIZATION HELPERS
# ==========================================

def format_board(state, policy=None):
    """
    Prints the Connect Four board (6x7).
    If policy is provided (length 7), prints the move probabilities above the board.
    """
    board_str = state.observation_string()
    rows = board_str.splitlines()

    print("\n--- BOARD ---")

    # Print move probabilities above the columns
    if policy is not None and len(policy) == NUM_ACTIONS:
        p_row = [f"{p:.2f}" for p in policy]
        print("  " + " | ".join(p_row) + "  <-- Move Probabilities")
        print("-" * (5 * NUM_COLS + 1))

    # Print the board itself (6 rows)
    for r in range(NUM_ROWS):
        # OpenSpiel's standard Connect Four board symbols are typically:
        # 0 | 1 | 2 | 3 | 4 | 5 | 6 (Player 0 / X)
        # . | . | . | . | . | . | .
        # The observation string usually has the symbols separated by spaces.

        # We need to ensure we only print the board symbols (X/O/.)
        row_chars = rows[r].strip().split()
        row_syms = [f" {char} " for char in row_chars]

        print(f"|{'|'.join(row_syms)}|")

    # Print column indices for human input
    print("-" * (5 * NUM_COLS + 1))
    col_indices = [f" {i} " for i in range(NUM_COLS)]
    print(f" {' '.join(col_indices)}")

def analyze_model_prediction(model, state, device):
    # This requires the ProZero model to be adapted for 6x7 input and 7 policy output

    obs = get_obs(state)
    x = torch.tensor(obs, dtype=torch.float32, device=device).unsqueeze(0)

    model.eval()
    with torch.no_grad():
        # Policy output (p) MUST now be of size 7, not 9
        p, v, r = model(x)

    # 1. Get Raw Probabilities (Sum = 1.0 over 7 columns)
    # The policy head p must output 7 logits for Connect 4 actions (columns 0-6)
    raw_probs = F.softmax(p, dim=-1)[0].cpu().numpy()

    # 2. Apply Legal Mask
    legal_mask = np.zeros(NUM_ACTIONS) # Size 7
    legal_mask[state.legal_actions()] = 1

    masked_probs = raw_probs * legal_mask

    # 3. Re-normalize (Sum = 1.0 over VALID actions)
    if masked_probs.sum() > 1e-6: # Check for near-zero sum
        masked_probs /= masked_probs.sum()
    else:
        # Fallback for safety
        print("Warning: Model output collapsed. Using uniform legal policy.")
        masked_probs = legal_mask / legal_mask.sum()

    v_probs = F.softmax(v, dim=-1)[0].cpu().numpy()

    print(f"\n[AI THOUGHTS]")
    print(f"Value Head (Win/Loss/Draw): {v_probs}")
    # Note: If v_fc2 outputs 1 (scalar value), change v[0][0] to v.item()
    print(f"Raw Evaluation Score: {v[0][0].item():.2f}")

    # Pass the MASKED probs (length 7) to the printer
    format_board(state, masked_probs)
    return masked_probs

# ==========================================
# 2. INTERACTIVE GAME
# ==========================================
def play_vs_ai(model_path, player=1):
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    # Init Model (ProZero class must be adapted for Connect 4 dimensions: 6x7, P=7)
    model = ProZero().to(device)
    try:
        model.load_state_dict(torch.load(model_path, map_location=device))
        print("Model loaded successfully!")
    except Exception as e:
        print(f"Could not load model. Error: {e}")
        return

    # CHANGE: Load Connect Four game
    game = pyspiel.load_game("connect_four")
    state = game.new_initial_state()

    print("\n" + "="*40)
    print(" HUMAN (X) vs ALPHAZERO (O) - Connect 4")
    print("="*40)
    print(f"Enter move index (0-{NUM_ACTIONS-1}):")
    print("0 1 2 3 4 5 6")

    while not state.is_terminal():
        current_player = state.current_player()
        player_symbol = 'X' if current_player == 0 else 'O'
        ai_symbol = 'O' if current_player == 0 else 'X'

        if current_player == player: # Human
            print(f"\nYour Turn ({player_symbol})")
            # Show AI's evaluation of the current state
            analyze_model_prediction(model, state, device)

            legal_moves = state.legal_actions()
            move = -1
            while move not in legal_moves:
                try:
                    move_str = input(f"Enter move {legal_moves} (Column 0-6): ")
                    move = int(move_str)
                    if move not in range(NUM_ACTIONS):
                        print(f"Invalid column. Must be between 0 and {NUM_ACTIONS-1}.")
                        move = -1
                except ValueError:
                    print("Invalid input. Please enter a number.")
                    continue
            state.apply_action(move)

        else: # AI
            print(f"\nAI Thinking ({player_symbol})...")
            probs = analyze_model_prediction(model, state, device) # Masked, re-normalized probs

            # Greedy selection (Best move according to the neural net policy)
            move = np.argmax(probs)
            print(f"AI chose column: {move}")
            state.apply_action(move)

    # Game Over
    print("\n" + "="*40)
    print("GAME OVER")
    format_board(state)
    returns = state.returns()
    winner = "Player X (0)" if returns[0] == 1.0 else ("Player O (1)" if returns[1] == 1.0 else "Draw")
    print(f"Winner: {winner}")
    print(f"Returns (P0, P1): {returns}")

if __name__ == "__main__":
    # Ensure you have a Connect Four-adapted ProZero model saved here
    # play_vs_ai("model_iter_connect4.pth", player=1) # Human is Player 1 (O)
    play_vs_ai("model_iter_100.pth",1)
```
